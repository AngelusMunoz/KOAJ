```python
# Revisar valores nulos

nulos = df.isnull().sum()
print(f"Nulos:\n{nulos[nulos > 0]}")
print(f"\nTotal filas: {len(df)}")
print(f"\nPorcentaje de nulos por columna:")
print((df.isnull().sum() / len(df) * 100).round(2))
```

```python
# Revisar valores duplicados

duplicados = df.duplicated(subset=['id_venta', 'numero_linea']).sum()
print(f"Duplpicados exactos: {df.duplicated().sum()}")
print(f"\nDuplicados por id_venta + numero_linea: {duplicados}")
```

```python
# Valores unicos por columna

columnas_categorias = ['empresa', 'ciudad', 'categoria', 'canal', 
                       'metodo_pago', 'temporada', 'genero_cliente', 
                       'tipo_cliente', 'tipo_prenda', 'departamento']

for col in columnas_categorias:
    print(f"{col}: {df[col].nunique()} valores únicos")
    print(f"  {sorted(df[col].dropna().unique())}\n")
```

```python
# Verificar los precios

print(f"Tipo de dato de precio: {df['precio_unitario'].dtype}")
print(f"\nPrimeros valores:\n{df['precio_unitario'].head(10)}")
print(f"\nRango: {df['precio_unitario'].min()} - {df['precio_unitario'].max()}")
```

```python
# Conversion de los precios para mejor visualizacion

df['precio_unitario'] = df['precio_unitario'].astype(str).str.replace('.', '', regex=False).str.replace(',', '.', regex=False).astype(float)

precios_negativos = df[df['precio_unitario'] < 0].shape[0]
print(f"\nPrecios negativos: {precios_negativos}")

print(f"\nRango: {df['precio_unitario'].min()} - {df['precio_unitario'].max()}")
df['precio_unitario']
```
